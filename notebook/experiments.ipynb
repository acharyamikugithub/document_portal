```python
from langchain_openai import OpenAIEmbeddings
emebedding=OpenAIEmbeddings(model="text-embedding-3-small")

```

```python
query="what is the capital of france?"
embed=emebedding.embed_query(query)
embed
```

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
file_path=os.path.join(os.getcwd(),"data", "attention.pdf")
```

```python
loader=PyPDFLoader(file_path)
```

```python
documents=loader.load()
```

```python
documents
```

```python
len(documents)
```

```python
## this is a experimental thing there is no deterministic way to split the text
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=150,
    length_function=len
)
```

```python
docs=text_splitter.split_documents(documents)
```

```python
len(docs)
```

```python
docs[0].metadata
```

```python
docs[0].page_content
```

```python
docs[0].metadata
```

```python
from langchain_community.vectorstores import FAISS
vectorestore=FAISS.from_documents(docs,emebedding)
```

```python
relevent_doc=vectorestore.similarity_search("llama2 finetuning benchmark experiments.",k=3)
```

```python
relevent_doc
```

```python
relevent_doc[1].page_content
```

```python
relevent_doc[2].page_content
```

```python
retriever=vectorestore.as_retriever()
```

```python
retriever.invoke("llama2 finetuning benchmark experiments")
```

```python
prompt_template="""
    Answer the question based on the context provided below.
    If the context does not contain sufficient information,respond with:
    "I do not have enough information about this"
    Context: {context}
    Question: {question}

    Answer"""
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context","question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
```

```python
rag_chain=prompt | llm | parser
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
rag_chain=(
    {"context":retriever | format_docs,"question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain.invoke("tell me about the llama2 finetuning benchmark experiments?")
```
